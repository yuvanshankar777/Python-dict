## Introduction

Dictionaries are a fundamental tool in data analysis for storing and managing structured information. They let you associate keys with values, enabling fast lookups, flexible data organization, and intuitive mapping of relationships in your datasets. Dictionaries are **unordered** — the items do not have a fixed sequence — and **mutable**, so you can add, update, or remove entries as your analysis evolves.

In Python, dictionaries are implemented using the `dict` class.

**Use cases:**

* Fast lookups by key
* Counting and grouping items
* Configuration settings and mappings

## Basic Properties

* **Mutability**: Mutable (can be changed after creation)
* **Ordered**: Yes (maintains insertion order)
* **Allows duplicate keys**: No (keys must be unique, values can be duplicated)
* **Indexing support**: No (access by key, not by position)
* **Keys must be**: Immutable types (strings, numbers, tuples)
* **Values can be**: Any data type

## Accessing Dict Objects

### Key

```python
person = {"name": "Alice", "age": 25, "city": "New York"}
```

```python
# Access using the same way as we do in lists/tuples/strings
# But we use keys instead of positions.
person["name"]
```

```python
person["age"]
```

```python
# Accessing non-existent keys would cause errors
person["hometown"]
```

### `get()` method

```python
city = person.get("city")
city
```

```python
# Non-existent key returns None when using this method
hometown = person.get("hometown")
print(hometown)
```

### Nested Dictionaries

```python
student = {
    "name": "Bob",
    "age": 20,
    "address": {
        "city": "Boston",
        "state": "MA",
        "zip": "02101"
    },
    "grades": {
        "Math": 85,
        "Science": 90
    }
}
```

```python
math_grade = student["grades"]["Math"]
print(math_grade) 
```

## Common Methods in `lst` class

### Adding Elements

```python
product_data = {
    'product_id': 'P101',
    'category': 'Electronics',
    'price': 299.99,
    'units_sold': 150,
    'rating': 4.5
}
```

```python
product_data
```

```python
product_data['description'] = 'ear buds'  # Adding a pair to a dictionary
```

```python
product_data
```

```python
# `update()` method: updates the dictionary with another dictionary
manufacturing_data = {'location': 'Taiwan', 'cost': 123.5}
product_data.update(manufacturing_data)
```

```python
product_data
```

### Removing Elements

```python
# `pop` method: removes a pair from the dict and returns the value. Key to be removed is specified.
location = product_data.pop("location")
location
```

```python
product_data  # Location is removed from the dictionary
```

### Accessing Methods

```python
# We can access only the keys using the `keys()` method
product_data.keys()
```

```python
# We can access only the values using the `values()` method
product_data.values()
```

```python
# We can access the key-value pairs using the `items()` method
product_data.items()
```

You can observe that the data type of these outputs are not lists or tuples but a unique object called `dict_items`.

You may learn more about them here: [<ins>`View Objects`</ins>](https://docs.python.org/3/library/stdtypes.html#dictionary-view-objects)
    

When you assign the `dict_keys` view of a dictionary to a variable, that variable **does not store a static copy of the keys**. Instead, it **references the original dictionary**. This means that if the dictionary’s contents change, the `dict_keys object` will **reflect those changes automatically**, even though you never modified the dict_keys object directly.

`dict_items` and `dict_values` works the same way.

```python
data = {'a': 1, 'b': 2}
keys_view = data.keys()  # dict_keys object
keys_view
```

```python
data['c'] = 3  # modifying the dictionary

keys_view  # Though, we didn't manually change the `keys_view`.
```

## Operators

### Merge (`|`)

```python
dict_1 = {"Manager": "Sundar", "Secretary": "Kanika"}
dict_2 = {"Engineer": "Nandan", "President": "Sudha"}

employee_dict = dict_1 | dict_2
employee_dict
```

```python
# If both dicts have common keys, second dict's key is considered as dict keys can't be identical.
dict_1 = {"Manager": "Sundar", "Secretary": "Kanika", "Engineer": "Keerthan"}
dict_2 = {"Engineer": "Nandan", "President": "Sudha"}

employee_dict = dict_1 | dict_2
employee_dict
```

### Equality (`==`)

```python
# Outputs True if both dicts have the same key-value pair no matter what the order is.
{'a': 'apple', 'c': 'cat'} == {'c': 'cat', 'a': 'apple'}
```

### Membership Operators (`in`, `not in`)

```python
operations = {'add': '+', 'subtract': '-', 'multiply': '*', 'division': '/'}
```

```python
# Check if a key exists in the `dict` object; use the `in` operator.
print("add" in operations)
```

```python
print("power" in operations)
```

```python
# Check if object doesn't exist; use the `not in` operator
print("floor_division" not in operations)
```

## Commonly Used Functions

```python
numerals  # `dict` object created at the top of this notebook.
```

```python
# `len` function outputs number of key-value pairs in `dict` object
len(numerals)
```

```python
# `max` function outputs the largest key in the `dict` object
max(numerals)
```

```python
# `min` function outputs the smallest key in the `dict` object
min(numerals)
```

```python
# `sum` function outputs the total sum of all the keys in the `dict` object provided keys are numbers
sum(numerals)
```

```python
# `sorted` function returns a new list that is a sorted version of keys of the original `dict` object
sorted(numerals)  # ascending order
```

```python
sorted(numerals, reverse=True)  # descending order
```

```python
# `type` function outputs the data type/class of the object
type(numerals)
```

```python
# `dir` function outputs the attributes and methods of the object
dir(numerals)
```

```python
# `id` function outputs the memory address of the object
id(numerals)
```

> **Note**:

> Merging operator (|) on dict objects produce new dict object. So, if you'd like to reuse the result again, bind it with a variable.

> Some methods directly changes the object in place and outputs `None`: for example, `update`, `clear`.

> Some methods outputs new objects without modifying the list: for example, `keys`, `values`.

> Some methods modifies the object and also outputs an object: for example, `pop`.

```python
delete_dict
```

```python
# Do the following to delete the whole dict from memory.
del delete_dict
```
